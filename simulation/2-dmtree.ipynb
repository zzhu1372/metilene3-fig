```python
for i in range(50):
    asize = 1+i%10
    
    cls = pd.read_table('./data/results/consensus/'+bg+'a10.LGGGBM.n40rep'+str(i)+'.metilene3/clusters.tsv')
    cls.index = cls['ID']
    cpgmet = met.loc[cls.index]
    cpgstd = cpgmet.std().sort_values()
    
    ### DMTree DMRs
    udmrs = pd.read_table('./data/results/consensus/'+bg+'a10.LGGGBM.n40rep'+str(i)+'.metilene3/DMRs-unsupervised.tsv', skiprows=2)
    udmrmet = []
    udmrs.loc[(udmrs['DMTree'].str.contains('P'))&\
                    (udmrs['#Hyper']>=2)&\
                    (udmrs['#Hypo']>=2)]['mean'].apply(lambda x:udmrmet.append(x.split('|')))
    udmrmet = pd.DataFrame(udmrmet).astype(float).T
    udmrmet.index = cls.index
    
    top1cent = sorted(cpgstd[cpgstd>cpgstd.quantile(0.99)].index)

    from sklearn.cluster import KMeans
    pca = PCA(n_components=2)
    pcaudmrmettmp = pca.fit_transform(np.array(udmrmet))
    pcatop1centtmp = pca.fit_transform(np.array(cpgmet[top1cent]))
    
    for k in range(3,8):
        cls['hc_dmt_k'+str(k)] = AgglomerativeClustering(n_clusters=k).fit_predict(udmrmet)
        cls['pcakmeans_dmt_k'+str(k)] = KMeans(n_clusters=k, random_state=0).fit_predict(pcaudmrmettmp)
        
        cls['hc_top1_k'+str(k)] = AgglomerativeClustering(n_clusters=k).fit_predict(cpgmet[top1cent])
        cls['pcakmeans_top1_k'+str(k)] = KMeans(n_clusters=k, random_state=0).fit_predict(pcatop1centtmp)

    for k in range(3,8):
        for method in ['hc_top1_k','pcakmeans_top1_k',\
                       'hc_dmt_k','pcakmeans_dmt_k']:
            consensus[method+str(k)][i] = consensus[method+str(k)].index.map(cls[method+str(k)])
    
    print(i)
```

```python
def cs(x,y):
    sumcs = 0
    sumall = 0
    import math
    for i,ii in enumerate(x):
        if math.isnan(x[i]) or math.isnan(y[i]): continue
        sumall+=1
        if x[i]==y[i]:
            sumcs+=1
    if sumall:
        return sumcs/sumall
    else:
        return None
```
